```python
import serial
import time
```

```python
# Replace 'COM3' with your ESP32 port
COM='COM8'
ser = serial.Serial(COM, 115200, timeout=1)
time.sleep(5)
```

```python
# "0b0010000100000000\n"  18-15
# "0b1000001000000000\n"  15-20
# "0b0001010000000000\n"  1-18

ch = "0b0001010000000000\n"

ser.write(ch.encode())

#time.sleep(1)
```
