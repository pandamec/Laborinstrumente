```python
from IPython.display import clear_output, display, HTML
import time

while True:
    # READ SENSOR HERE
    T = controller.getTemperature()

    clear_output(wait=True)

    display(HTML(f"""
    <div style="
        font-size: 150px;
        color: blue;
        font-weight: bold;
        text-align: center;
    ">
        T = {T:.3f} K
    </div>
    """))

    time.sleep(0.1)
```
