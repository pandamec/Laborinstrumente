```python
ic = ctypes.cdll.LoadLibrary(r"C:\TUC\03 Scripts\Laborinstrumente\src\DIC\tisgrabber_x64.dll")
tis.declareFunctions(ic)
ic.IC_InitLibrary(0)
```

```python
hGrabber = ic.IC_LoadDeviceStateFromFile(None,tis.T(r"C:\TUC\03 Scripts\Laborinstrumente\src\DIC\device_for_dic_2656x2296_bin_temp.xml"))
```

```python
# hGrabber = ic.IC_ShowDeviceSelectionDialog(None)
# ------------------------------
# --------- live video ---------
# ------------------------------
# hGrabber = ic.IC_LoadDeviceStateFromFile(None,tis.T("device_for_dic_2560x1920.xml"))


if(ic.IC_IsDevValid(hGrabber)):
   # ic.IC_SetVideoFormat(hGrabber, tis.T("Y16 (5320x4600)"))
   #ic.IC_SetVideoFormat(hGrabber, tis.T("Y16 (2656x2296) [Binning 2x]"))
   ic.IC_StartLive(hGrabber,1)

   flag = True
   counter = 0
   wait_frame_ms = 2000
   wait_per_image_s = 60
   max_imgs = 1000

   while flag:

      counter+=1
      ic.IC_SnapImage(hGrabber, wait_frame_ms)
      ic.IC_SaveImage(hGrabber, tis.T(f"test_DIC\\260713_test{counter:04d}.bmp"),
                      tis.ImageFileTypes['BMP'], 90)
      #ic.IC_SaveImage(hGrabber, tis.T("260713DIC\\260713_output_imgs\\240704_Si_Al_RT_to_lowT_{counter:04d}.bmp"),
      #                tis.ImageFileTypes['BMP'], 90)
      if counter == max_imgs:
         flag = False
      time.sleep(wait_per_image_s - wait_frame_ms / 1000)

      print("Picture taken")



   ic.IC_StopLive(hGrabber)
else:
   ic.IC_MsgBox(tis.T("No device opened"), tis.T("Simple Live Video"))

ic.IC_ReleaseGrabber(hGrabber)

```
