```python
def parse_sourcemeter_data(data_str):
    parts = data_str.split(',')
    
    rows = []
    
    for i in range(0, len(parts), 3):
        r_raw = parts[i]
        t_raw = parts[i+1]
        c_raw = parts[i+2]
        
        # Remove units by replacing strings
        r = float(r_raw.replace('OHM', ''))
        t = float(t_raw.replace('SECS', ''))
        c = float(c_raw.replace('ADC', ''))
        
        rows.append((r, t, c))  # tuple: (resistance, time, current)
    
    return rows
```

```python
import pandas as pd
a=parse_sourcemeter_data(read)
print(pd.DataFrame(a))
```
